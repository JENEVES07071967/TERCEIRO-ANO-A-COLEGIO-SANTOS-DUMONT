```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


# Imprime a mediana das notas do array
print(np.array([2.5] * 10).mean())

# Cria um array de dez posições, todas ocupadas pela nota 2.5
np.array([2.5] * 10)

# Cria um array de dez posições, todas ocupadas pela nota 3.5
np.array([3.5] * 10)

# Com a função append(), juntaremos esses dois arrays em um único, que atribuiremos a uma variável filme1

filme1 = np.append(np.array([2.5] * 10), np.array([3.5] * 10))

# Para outra variável, filme2, também atribuiremos dois conjuntos de notas — dessa vez, 5 e 1, também repetidas dez vezes cada uma

filme2 = np.append(np.array([5] * 10), np.array([1] * 10))

# Imprimindo a média desses dois filmes

print("Média dos dois filmes")
print(filme1.mean(), filme2.mean())
print()

# Imprimindo a mediana desses dois filmes

print("Mediana dos dois filmes")
print(np.median(filme1), np.median(filme2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Adiciona uma coluna 'filme' em cada DataFrame para identificação
notas_do_toy_story['filme'] = 'Toy Story'
notas_do_jumanji['filme'] = 'Jumanji'

# Combina os dois DataFrames em um único
dados_combinados = pd.concat([notas_do_toy_story, notas_do_jumanji])

# Plota a distribuição usando displot()
# O 'hue' usa a nova coluna 'filme' para separar os gráficos por cor
sns.displot(
    data=dados_combinados,
    x='rating',
    hue='filme',
    kind='hist',
    kde=True,
    stat='density',
    common_norm=False,
)

# Adiciona um título ao gráfico
plt.title('Comparação da Distribuição de Notas dos Filmes')
plt.show()
```
